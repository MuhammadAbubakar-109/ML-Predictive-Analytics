```python
import sys; sys.path.append('..')
import pandas as pd, matplotlib.pyplot as plt
raw = pd.read_csv('../data/raw/Loan_Default.csv', low_memory=False)
raw['Status'].value_counts(normalize=True)
```

## Leakage check
These columns are missing almost only for defaulted loans -> excluded in `configs/config.yaml`.

```python
cols = ['rate_of_interest','Interest_rate_spread','Upfront_charges','LTV','property_value','dtir1','income']
pd.DataFrame({c: {'default rate if missing': raw.Status[raw[c].isna()].mean(), 'default rate if present': raw.Status[raw[c].notna()].mean()} for c in cols}).T
```

```python
raw.groupby('Credit_Worthiness')['Status'].mean().plot.bar(); plt.show()
```

```python
raw.groupby(pd.cut(raw.Credit_Score,[499,600,700,800,900]))['Status'].mean().plot.bar(); plt.show()
```
